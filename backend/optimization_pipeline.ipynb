```python
def optimization_pipeline(event,budget,directory='',x_b_fix=False):
    
    '''This code runs the optimization pipeline which is called from the dashboard. The function sets up the optimization parameters,
    calls AMPL, and runs the post-processing'''

    '''Inputs: 
    event: 
    budget: budet (in $ million)
    fix_constraint: set to 1 if we want to fix the the building stock or hardening scenario. For example, if we run the optimization for one event,
    then want to evaluate the impacts of a different event, we can fix the output of the first run and run the second event
    x_b_fix: the fixed building stock matrix [neighborhood, archetype]
    '''
    budget = budget*10**6 # set budget to millions
    
    # ============================ SET UP ===========================================================================
    
    print('Setting up optimization...')
    
    set_up_optimization(event,budget,directory=directory)
    
    
    # =========================== Run optimization ====================================================================

    print('Running optimization...')
    
    run_opt(budget,directory=directory,x_b_fix=x_b_fix);
    
    # ========================== Post-process results ================================================================
    
    post_process_results()
    
    # ========================= Generate outputs ====================================================================
      
    print('Optimization complete!')
    
```
